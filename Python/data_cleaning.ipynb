```python
from pathlib import Path
import pandas as pd

project_folder = Path(r"E:\Projects\SQL\Project 2\raw data")

datasets = {}

for file in project_folder.glob("*.csv"):
    datasets[file.stem] = pd.read_csv(file)

print(datasets.keys())
```

```python
books = datasets["books"]
branch = datasets["branch"]
employees = datasets["employees"]
issued_status = datasets["issued_status"]
members = datasets["members"]
return_status = datasets["return_status"]
```

```python
for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(f"{name.upper()}")
    print(f"{'=' * 50}")
    print("Row count:", len(df))
    print("Missing values:")
    print("\nData types:")
    print(df.dtypes)
    print(df.isnull().sum())
```

```python
primary_keys = {
    "books": "isbn",
    "branch": "branch_id",
    "employees": "emp_id",
    "issued_status": "issued_id",
    "members": "member_id",
    "return_status": "return_id"
}

for table, key in primary_keys.items():
    df = datasets[table]

    print(
           f"{table}.{key}: "
           f"{df[key].nunique()} unique / "
           f"{len(df)} total"
       )
```

```python
datasets["issued_status"]["issued_date"] = pd.to_datetime(
    datasets["issued_status"]["issued_date"]
)

datasets["members"]["reg_date"] = pd.to_datetime(
    datasets["members"]["reg_date"]
)

datasets["return_status"]["return_date"] = pd.to_datetime(
    datasets["return_status"]["return_date"]
)
```

```python
datasets["branch"]["contact_no"] = (
    datasets["branch"]["contact_no"]
    .astype(str)
)
```

```python
from pathlib import Path

cleaned_folder = project_folder / "cleaned"
cleaned_folder.mkdir(exist_ok=True)

for name, df in datasets.items():
    output_path = cleaned_folder / f"{name}.csv"
    df.to_csv(output_path, index=False)
    print(f"Saved: {output_path}")
```

```python
for file in cleaned_folder.glob("*.csv"):
    df = pd.read_csv(file)

    print(f"{file.name}: {len(df)} rows")
```
